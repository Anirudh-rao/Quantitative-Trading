# A Map of Trading Strategies

There is no single way to trade quantitatively. There is a whole landscape of strategy families, each exploiting a different reason the market pays you. This lesson is a map of that landscape. Knowing it helps you place your own ideas, understand what you're competing against, and — crucially — judge honestly which approaches are realistic for an individual versus a well-funded firm.

By the end of this lesson you will be able to:

- Name the major families of trading strategies and the edge each exploits
- Place a strategy on the spectrum from high-frequency to long-term
- Estimate the capital, data, and infrastructure each family demands
- Judge which strategies are realistic for an individual versus a firm


## 1. The Big Idea: Every Strategy harvests a specific edge

Recall that an edge is a reason the market pays you. Strategy families are simply organized around different reasons. Some exploit that prices trend; others that they snap back; others that someone must pay a premium to offload risk. When you understand the underlying edge, the family's behavior — its hit rate, its risk, its capacity — follows naturally.

A theme worth holding onto: each edge exists because someone on the other side is willing to lose to you on average, and the family's risk profile is just the shape of how that counterparty occasionally wins big. Momentum's slow institutional buyers usually feed you, but a sharp reversal hands the loss back. Mean reversion's panicked sellers usually overshoot, but sometimes the panic was justified and the price keeps falling. There is no free edge; every family is "get paid most of the time, occasionally get hurt," and the families differ mainly in when and how badly the hurt arrives.

## 2. The Major Families

### a. Trend/Momentum

The edge: prices that have moved in one direction tend to keep moving, because investors react slowly to news and pile in over time. You buy strength and sell weakness. Momentum strategies typically have a **low hit rate but high payoff ratio** — many small losses, a few large wins that pay for everything.They work across stocks, futures, and currencies and can run from days to months. The main risk is sharp reversals that turn a string of winners into losses.

Because the payoff is so skewed toward rare big winners, the psychological challenge of trend following is enduring long strings of small losses without abandoning the system right before the move that pays for them all. The math from the expectancy lesson applies directly: a 38% hit rate with a payoff ratio above 2 is comfortably profitable, but it feels like losing, and most people quit.


### b. Mean Reversion

The edge: prices that move too far from a "normal" level tend to snap back, often because of temporary forced buying or selling. You buy what's fallen and sell what's risen. Mean reversion usually has a **high hit rate but small payoff ratio** — many small wins and the occasional large loss when the move doesn't revert (a falling knife). It needs careful risk control precisely because of those rare but severe failures.

The defining danger is that mean reversion and a genuine regime change look identical at the moment you enter. A stock down 8% might be an overreaction that bounces, or the first day of a collapse on news you haven't processed. Because the losses are rare but large, a mean-reversion trader who skips stop-losses or oversizes can give back months of small wins in a single trade. The high hit rate is seductive in exactly the dangerous way the expectancy lesson warned about.


### c. Statistical Arbitrage

The edge: relationships between assets are more stable than the assets themselves. The classic example is **pairs trading**: two historically linked stocks drift apart, so you short the rich one and buy the cheap one, betting the spread reconverges. This is mean reversion applied to relationships rather than single prices. It demands solid statistics, good data, and many simultaneous positions to diversify the idiosyncratic risk of any single pair.

The appeal is that a well-constructed spread is roughly **market-neutral**: if both stocks fall together, the long and short offset, so you're betting only on the relationship, not on market direction. The catch is that relationships break — a merger, a scandal, or a changed business model can permanently sever a pair that was linked for a decade. Diversifying across many pairs is not optional; it's the only thing standing between you and a single broken relationship taking an outsized loss.

### d. Cross-Sectional Factors

The edge: certain measurable characteristics predict relative returns across a large universe of assets. Instead of timing the market, you rank every stock by a factor — value (cheap vs expensive), momentum, quality, size — and go long the top, short the bottom. This is the foundation of systematic equity investing. It needs a broad universe and clean fundamental data, but trades slowly and scales to large capital.

The power here is **diversification across names**: any single stock can do anything, but the average behavior of the top-ranked hundred versus the bottom-ranked hundred is far more stable. You're not predicting individual companies; you're harvesting a tilt that holds across a crowd. This is why factor strategies can run enormous capital and rebalance only monthly. The risk is that popular factors can underperform for years — value famously did through much of the 2010s — testing the patience of anyone running them.

### e. Carry

The edge: you get paid simply for holding something others want to avoid. Examples: earning the interest-rate differential between two currencies, or the roll yield in futures markets. Carry strategies tend to make steady money for long stretches and then suffer sudden, sharp losses when the risk everyone was avoiding actually materializes. The profile is "pick up nickels, occasionally get hit by a truck" — so risk sizing is everything.

The deep insight is that carry is compensation for bearing a risk, so a smooth, attractive carry return is not free money — it is rent you collect for standing in front of an occasional disaster. The 2008 unwinding of currency carry trades and various "volatility selling" blowups are the trucks. If a carry strategy looks like it never loses, you are simply being paid before the bill arrives, and the right response is to size as if the truck is coming, because eventually it is.


### f. Market Making

The edge: you quote both a buy and a sell price and earn the spread between them, providing liquidity to others. You don't bet on direction; you profit from turnover while managing the inventory you accumulate. This is a volume game requiring extremely low latency, deep technical infrastructure, and tight risk controls. It is dominated by specialized firms and is generally not realistic for individuals.


### g. Event-Driven

The edge: predictable price behavior around scheduled or announced events — earnings releases, index additions, mergers. Merger arbitrage, for example, buys a takeover target trading below the offer price and earns the gap if the deal closes, while risking a loss if it breaks. These strategies need event data and a clear understanding of the specific situation's risks.


### h. Classicial Arbitrage

The edge: the same thing trades at two different prices in two places, so you buy low and sell high simultaneously for a near risk-free profit. True arbitrage is rare, tiny, and fiercely contested — it's gone in milliseconds. In practice it belongs to the fastest, best-capitalized firms. For everyone else, "arbitrage" usually means statistical arbitrage, which carries real risk.



## 3. Time Horizons

The same edge behaves very differently depending on how fast you trade. Horizon shapes everything: cost sensitivity, data needs, and who you compete against.

- **High-frequency (HFT)** — sub-second to seconds. Tiny edges per trade, enormous trade counts. Requires colocated servers and specialized engineering. Off-limits to individuals.
- **Intraday** — minutes to hours, flat by the close. Sensitive to costs and execution; needs good intraday data.
- **Swing** — days to weeks. The sweet spot for many individuals: meaningful edges, manageable data and infrastructure, costs that don't dominate.
- **Long-term** — weeks to years. Low trading costs, capacity for large capital, but you wait a long time to learn whether your edge is real.

A general rule: **the faster you trade, the more costs and infrastructure matter, and the more you compete directly with professional firms**. Slower horizons are where individuals have the most room.

There is a second, less obvious consequence of horizon: it determines how quickly you can learn. A strategy that trades 50 times a day produces thousands of data points a year, so you discover whether the edge is real within months. A strategy that rebalances monthly might take a decade to accumulate enough trades to distinguish skill from luck — the sample-size problem from the expectancy lesson, now applied to your whole career. Faster horizons are statistically easier to validate but harder to win at; slower horizons are easier to win at but agonizingly slow to confirm. Many individuals land on swing trading precisely because it balances these two pressures.

## 5. Individual versus Firms: An honest assessment

Be clear-eyed about where you can compete. A firm can spend millions on colocated hardware, buy exclusive datasets, hire teams of PhDs, and employ thousands of dollars of capital per strategy. You cannot beat them at their own game — speed, data exclusivity, and raw scale.

But you have advantages they don't. You can trade strategies too small to interest a large fund (capacity-constrained edges they must ignore). You answer to no investors, so you can sit patiently in a slow strategy through a flat year. And you have far lower overhead.

The practical takeaway: as an individual, focus on **slower horizons** (swing to long-term) and **lower-infrastructure families** (trend, mean reversion, cross-sectional factors, careful carry). Leave market making and true arbitrage to the firms — competing there is a guaranteed way to lose money to faster, better-funded opponents.

>Your edge as an individual is not speed or scale. It's patience, small size, and the freedom to trade what big players can't be bothered with.

The "too small to matter" point deserves a number. A fund managing $5 billion cannot meaningfully deploy capital into an edge that only works on $2 million of a thinly traded instrument — even if the edge returns 40% a year, the $800,000 of profit is a rounding error against their costs, and trying to size up would destroy the edge through market impact. For you, that same $2 million (or $20,000) edge is a real opportunity. Whole categories of edge are abandoned by professionals purely because they don't scale, and that abandoned territory is where an individual actually has a fighting chance.

## 6. A Worked Example : Placing one idea

Suppose your hypothesis is: *"Stocks that rose the most over the past 12 months keep outperforming over the next month." Where does it sit on the map?*

- Family: cross-sectional momentum — you rank a universe and go long winners, short losers.
- Horizon: monthly rebalance, so long-ish swing. Low trading cost sensitivity.
- Data: daily prices for a broad universe, survivorship-adjusted.
- Infrastructure: modest — you can rebalance once a month from a script.
Realistic for an individual? Yes, at small scale.


In two minutes the map told you this is a viable idea for an individual, what data you need, and roughly how often you'll trade. That is exactly what the map is for.


## 7. A Second Worked Example: the same idea , wrong family

Now watch the map catch a mistake. Suppose you love the feel of the breakout charts you see online and decide: *"I'll buy crypto altcoins the instant they spike 20% on a one-minute candle and scalp the continuation." Where does that sit?*

- Family: intraday momentum / scalping — chasing very short-term continuation.
- Horizon: seconds to minutes. Brutally cost- and latency-sensitive.
- Data: real-time tick or one-minute data across many venues.
- Infrastructure: low-latency execution, robust automation, fast risk controls.
- Realistic for an individual? Barely, and you're competing with market makers and HFT firms who see the same spike microseconds before you and have already traded it.

Same instinct (momentum), but a horizon that drags you straight into the firms' home turf. The map's verdict is honest and uncomfortable: the edge may be real, but you are the slowest, most expensive participant in that arena. A wiser move is to express the *same momentum instinct on a daily horizon*, where your latency disadvantage stops mattering. The map didn't just classify the idea — it told you to change the horizon, not the instinct.


## 8. Common Mistakes

- Chasing the glamorous families. Beginners gravitate to HFT and arbitrage — precisely the areas where individuals cannot compete.
- Ignoring capacity. A strategy that works with $10,000 may stop working at $10 million, and vice versa. Match the strategy to your capital.
- Underestimating data and cost demands. Faster strategies live or die on data quality and execution costs that beginners forget to budget for.
- Forcing one family onto every market. A momentum mindset applied to a mean-reverting instrument produces consistent losses.
- Confusing statistical arbitrage with risk-free arbitrage. "Stat arb" can and does lose money; only classical arbitrage is (briefly) near risk-free.
